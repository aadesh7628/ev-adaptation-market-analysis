```python
from pyspark.sql.functions import current_timestamp

source_table = "ev_registration.default.vehicle_registrations_500_k"
bronze_table = "ev_registration.bronze.bronze_vehicle_registrations"
```

```python
vehicle_df = spark.read.table(source_table)
```

```python
bronze_df = (
    vehicle_df
    .withColumn("ingestion_timestamp", current_timestamp())
)
```

```python
bronze_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(bronze_table)
```

```python
bronze_df.show(10)
```
